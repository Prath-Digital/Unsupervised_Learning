```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
import pickle
import os

# Load data
data_path = os.path.join('..', 'data', 'hybrid_recommendation_10000.csv')
df = pd.read_csv(data_path)

# Preprocess: Encode categorical
df['item_category'] = df['item_category'].astype('category')
df_encoded = pd.get_dummies(df, columns=['item_category'])

# Features and Target
X = df_encoded.drop(['user_id', 'item_id', 'purchased'], axis=1)
y = df_encoded['purchased']

# Train Random Forest
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)

# Save model
model_data = {
    'model': model,
    'features': X.columns.tolist(),
    'df_template': df.head(1).drop(['user_id', 'item_id', 'purchased'], axis=1, errors='ignore')
}

with open('model.pkl', 'wb') as f:
    pickle.dump(model_data, f)

# Save metrics
metrics = {
    'num_samples': len(df),
    'avg_rating': df['item_rating'].mean(),
    'purchase_rate': df['purchased'].mean()
}
with open('metrics.pkl', 'wb') as f:
    pickle.dump(metrics, f)

print("Hybrid training completed successfully.")
```
