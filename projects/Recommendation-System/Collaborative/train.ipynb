```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
import pickle
import os

# Load data
data_path = os.path.join('..', 'data', 'collaborative_user_item_10000.csv')
df = pd.read_csv(data_path)

# Preprocess: User-item matrix
user_item_matrix = df.pivot(index='user_id', columns='item_id', values='rating').fillna(0)

# Train NearestNeighbors (Item-Item Collaborative Filtering)
model = NearestNeighbors(metric='cosine', algorithm='brute')
model.fit(user_item_matrix.values.T) # Fit on items

# Save model
model_data = {
    'model': model,
    'user_item_matrix': user_item_matrix,
    'item_ids': user_item_matrix.columns.tolist()
}

with open('model.pkl', 'wb') as f:
    pickle.dump(model_data, f)

# Save metrics
metrics = {
    'num_users': user_item_matrix.shape[0],
    'num_items': user_item_matrix.shape[1],
    'sparsity': 1.0 - (df.shape[0] / (user_item_matrix.shape[0] * user_item_matrix.shape[1]))
}
with open('metrics.pkl', 'wb') as f:
    pickle.dump(metrics, f)

print("Collaborative training completed successfully.")
print(f"Users: {user_item_matrix.shape[0]}, Items: {user_item_matrix.shape[1]}")
```
