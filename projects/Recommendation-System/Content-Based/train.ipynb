```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pickle
import os

# Load data
data_path = os.path.join('..', 'data', 'content_based_products_10000.csv')
df = pd.read_csv(data_path)

# Preprocess: combine features into a single string for TF-IDF
df['combined_features'] = (
    df['category'].astype(str) + " " +
    df['subcategory'].astype(str) + " " +
    df['brand'].astype(str) + " " +
    df['feature_eco'].astype(str) + " " +
    df['feature_premium'].astype(str) + " " +
    df['feature_portable'].astype(str) + " " +
    df['feature_beginner'].astype(str)
)

tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['combined_features'])

# Calculate similarity
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# Save model
model_data = {
    'cosine_sim': cosine_sim,
    'df': df,
    'item_indices': pd.Series(df.index, index=df['item_name']).to_dict()
}

with open('model.pkl', 'wb') as f:
    pickle.dump(model_data, f)

# Save metrics
metrics = {
    'avg_similarity': np.mean(cosine_sim),
    'num_products': len(df)
}
with open('metrics.pkl', 'wb') as f:
    pickle.dump(metrics, f)

print("Content-Based training completed successfully.")
print(f"Products: {len(df)}")
```
